```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
insurance_df=pd.read_csv("./insurance.csv")
insurance_df.head()
```

```python
insurance_df.isnull().sum()
```

**checking for outliers**

```python
d=dict(insurance_df.dtypes)
d
```

```python
numerical_cols=[]
for i in d:
    if d[i]!="O":
        numerical_cols.append(i)
numerical_cols

```

```python
plt.figure(figsize=(12, 6))

for i, col in enumerate(numerical_cols, 1):
    plt.subplot(2, 4, i)
    plt.boxplot(insurance_df[col], vert=False)
    plt.title(col)

plt.tight_layout()
plt.show()
```
